```python
# Import needed libraries for text vectorization
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
# You can use your cleaned corpus or simple test sentences
sentences = [
    "I agree with you."
    "That sounds like a good good good idea."
    "I'm not so sure about that, though." 
]

sentences
```

## TF counts how many times each word appears.
### Example: "did did it" → did=2, it=1


```python
# Term Frequency = raw counts of words in each sentence
tf_vectorizer = CountVectorizer()

# Fit and transform the sentences
tf_matrix = tf_vectorizer.fit_transform(sentences)

# Convert to DataFrame for easy viewing
pd.DataFrame(tf_matrix.toarray(), columns=tf_vectorizer.get_feature_names_out())
```

### Binary = ignores word counts.
### If a word appears at least once → 1, else → 0.


```python
# Binary Vectorization turns counts into 0/1 (present or not present)
binary_vectorizer = CountVectorizer(binary=True)
binary_matrix = binary_vectorizer.fit_transform(sentences)

pd.DataFrame(binary_matrix.toarray(), columns=binary_vectorizer.get_feature_names_out())
```

### L1 normalization: divide counts by total words in the sentence.
### Useful to compare long vs short sentences fairly.


```python
# TF with L1 normalization → values become proportions that sum to 1
tf_l1 = TfidfVectorizer(use_idf=False, norm='l1')
tf_l1_matrix = tf_l1.fit_transform(sentences)

pd.DataFrame(tf_l1_matrix.toarray(), columns=tf_l1.get_feature_names_out())
```

### TF-IDF = TF * IDF
### Rare words get higher weight, common words get lower weight.


```python
# Raw TF-IDF before applying L2 normalization
tfidf_raw = TfidfVectorizer(norm=None)   # norm=None → no normalization
tfidf_raw_matrix = tfidf_raw.fit_transform(sentences)

pd.DataFrame(tfidf_raw_matrix.toarray(), columns=tfidf_raw.get_feature_names_out())
```

### L2 normalization scales sentence vectors to length = 1.
### Best for cosine similarity and ML models.
### Most NLP pipelines use TF-IDF + L2.


```python
# This is the most commonly used text representation in ML.
tfidf_l2 = TfidfVectorizer(norm='l2')    # default and recommended
tfidf_l2_matrix = tfidf_l2.fit_transform(sentences)

pd.DataFrame(tfidf_l2_matrix.toarray(), columns=tfidf_l2.get_feature_names_out())
```

### Vocabulary = dictionary of word:index mapping.
### Used internally for converting words to vector positions.


```python
# Vocabulary mapping → each word gets a unique index
tfidf_l2.vocabulary_
```

```python
# Combine all representations for comparison
comparison = pd.DataFrame({
    "Method": ["TF", "Binary", "TF + L1", "TF-IDF Raw", "TF-IDF L2"],
    "Description": [
        "Raw word counts",
        "0/1 presence",
        "TF normalized to sum=1",
        "No normalization",
        "Standard TF-IDF with L2"
    ]
})

comparison
```
